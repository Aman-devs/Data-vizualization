```python
import seaborn as sns
import pandas as pd
import cufflinks as cf
from plotly.offline import iplot
cf.go_offline()
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
tips['total_bill'].dtype
```

```python
import plotly.express as px

fig = px.line(tips, y='total_bill', title='Total Bill Distribution')
fig.show()
```

```python
fig = px.bar(tips.groupby('day')['tip'].mean().reset_index(), x='day', y='tip', title='Total Bill Distribution')
fig.show()
```

```python
fig=px.scatter(tips,x='total_bill',y='tip',color='day',title='Total Bill vs Tip')
fig.show()
```

```python
fig=px.box(tips[['total_bill','day']],y='day',x='total_bill',title='Total Bill Distribution by Day')
fig.show()
```
